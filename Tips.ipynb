1. Generate the docker file. In this example ./Dockerfile

2. Build the Docker image 

```python
docker build -t guokai94/mpcpy:latest -f .\Dockerfile
```

3. Login and push (mylogin account qq mail)

```python
docker login
```

4. Upload to the cloud

```python
docker push guokai94/mpcpy:latest
```

5. Go to Myriad: pull the container to Myriad

```python
# Load the Apptainer module
module load apptainer

apptainer build --docker-login -sandbox mpcpy.sif docker://guokai94/mpcpy:latest
```

6. Create a Writable Directory on Host:

```python
mkdir -p ~/mpcpy_writable_tmp
chmod 777 ~/mpcpy_writable_tmp
```

Execution: Run a specific command within container

```python

# Define the path to the container and the Python script
CONTAINER_PATH=~/mpcpy.sif

# Define the directory for the GitHub repository
REPO_DIR=~/mpcpy_writable_tmp/202405_GB_model

# Get the parameter file corresponding to this task
PARAM_FILE=$(sed -n "${SGE_TASK_ID}p" /home/ucbqgch/mpcpy_writable_tmp/202405_GB_model/Code/param_list.txt)

# Check if PARAM_FILE is set
if [ -z "$PARAM_FILE" ]; then
    echo "Error: PARAM_FILE is empty. Please check your param_list.txt and ensure it is correctly formatted."
    exit 1
fi

# Run the script inside the container with binding to writable temporary directory
apptainer exec --bind ~/mpcpy_writable_tmp:/tmp $CONTAINER_PATH ~/identification.sh "$PARAM_FILE"
```
